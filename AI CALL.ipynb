```python
from google import genai
from google.genai import types
from PIL import Image
from io import BytesIO
import os
from dotenv import load_dotenv
import requests

load_dotenv()   # reads .env into environment for local dev

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
print(GEMINI_API_KEY)




# The client gets the API key from the environment variable `GEMINI_API_KEY`.
userinput = input()
client = genai.Client(api_key=GEMINI_API_KEY)

response = client.models.generate_content(
    model="gemini-2.5-flash", contents="Create an infographic where a fun, gimmicky character or mascot represents **"+ userinput + "** and explains it step by step. Guidelines: * The mascot should embody **"+ userinput + "** in a playful, recognizable way (e.g., if the topic is gravity, the mascot could be a little planet character). * The infographic should be divided into 3–5 clear steps or sections. * Each step should use **visual storytelling** with the mascot demonstrating, pointing, or acting out the concept. * **Use the minimum amount of text needed** — only short labels or simple captions where visuals alone are unclear. * Style: clean, cartoony, colorful, and easy to understand. Structure: 1. Title area: introduce the mascot and the concept (**"+ userinput + "**). 2. Step 1: Mascot shows the starting point or problem. 3. Step 2: Mascot demonstrates how **"+ userinput + "** works. 4. Step 3: Mascot shows the result, effect, or why it matters. 5. Closing: Mascot gives a final “thumbs up” or summary pose.")
print(response.text)

response2 = client.models.generate_images(
    model='imagen-4.0-generate-001',
    prompt=response.text,
    config=types.GenerateImagesConfig(
        number_of_images= 1,
    )
)
for generated_image in response2.generated_images:
  generated_image.image.show()
```
